Concatenate = Songjokto kora

```python
#np.concatenate((a,b), axis = 0/1)-> 0 = row, 1 = column
con_row = np.concatenate((a,b), axis = 0) # row wise

print(con_row)

con_col = np.concatenate((a,b), axis = 1) # row wise

print(con_col)
```

```python
#array split
a = np.random.randint(1,10, size = (10,))

# array_split function when random division
splitted_array = np.array_split(a,3)
print(splitted_array)

#split function -> when equal division
splitted_array = np.split(a,3)
```

Arithmetic operator and mathematical functions

```python
#faster for vectorization

import numpy as np
x = np.array([ 10, 20,30,40])
y = np.array([2,3,4,5])

add = x+y
print(add)

sub = x-y
print(sub)

mul = x*y
print(mul)

div = x/y
print(div)

reminder = x % y
print(reminder)
```

```python
#Mathematical function

#trigonometry

#sin, cos, tan takes values in radian

sin_val = np.sin(x)
print(sin_val)

cos_val = np.cos(x)
print(cos_val)

deg_con = np.rad2deg(x)
print(deg_con)
```

```python
#log

print(x)

base_10_log_val = np.log10(x)
print(base_10_log_val)

two_diye_bhag = np.log2(x)
print(two_diye_bhag)


#Squere root

sqrt = np.sqrt(x)
print(sqrt)
```

```python
#different away
s = np.sum(x)
print(s)

prefix_sum
cumul = np.cumulative_sum(x)
print(cumul)
```

Logical Function


```python
# comparison - >, <, >=, <=, ==, !=

greater_then = x>y
print(greater_then)

equal = x==y
print(equal)

# all() -> Shob true kina
# any() -> jekono akta true kina

print(np.all(greater_then))
print(np.any(equal))
```

Sorting

```python
#inplace
x = np.array([10,2,16,100])

z = x.copy()

print(z)
z.sort()
print(z)
```

```python
# copy sorting

print(x)
sort_arr = np.sort(x)

print(sort_arr)
print(x)
```

```python
#2D array sorting
mat = np.array([[10,3,5],[8,4,9]])

# horizon axis = 1
hor_sort = np.sort(mat, axis=1)
print(hor_sort)

# Vertical axis = 0

vert_sort = np.sort(mat,axis = 0)
print(vert_sort)
```
