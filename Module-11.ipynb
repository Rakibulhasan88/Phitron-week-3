```python
import numpy as np
#np.random.randint(start number, end number, size = (row,column))
arr = np.random.randint(1,100, size=(10,5)) # sample-10, feature-5
arr
```

```python
arr.shape
```

```python
b = arr.reshape(5,10)
b.shape

b.ndim
b
```

```python
# make 1D array
flatten = b.flatten()
print(flatten.ndim)
flatten

column_wise_flattening = np.ravel(b, order = 'C')
print(column_wise_flattening)
```

```python
a = np.random.randint(1,10, size = (2,3))
b = np.random.randint(20,30,size = (2,3))
print(a)
print(b)
```

Concatenate = Songjokto kora

```python
#np.concatenate((a,b), axis = 0/1)-> 0 = row, 1 = column
con_row = np.concatenate((a,b), axis = 0) # row wise

print(con_row)

con_col = np.concatenate((a,b), axis = 1) # row wise

print(con_col)
```

```python
#error

#row wise concatenate  korar somoy column same thakbe
#col wise concat korar somoy row ta same thakte hobe

x = np.random.randint(1,10, size = (2,3))
y = np.random.randint(1,10,size = (1,3))

r = np.concatenate((x,y), axis = 1)
print(r)
```

Transpose

```python
#Transpose = make row to column

mat = np.array([[10,20,30],
                [30,40,50]
                ])

transpose = mat.T
print(transpose)
```

```python
#array split
a = np.random.randint(1,10, size = (10,))

# array_split function when random division
splitted_array = np.array_split(a,3)
print(splitted_array)

#split function -> when equal division
splitted_array = np.split(a,3)
```
