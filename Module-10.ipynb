Ndarray

```python
import numpy as np

# 1d array

arr1 = np.array([1,2,3,4,5])

print(type(arr1))
print(arr1.ndim)

# 2d Array

arr2 = np.array([[1,2,3], #row1
                 [4,5,6] #row2
                ]) 

print(arr2.ndim)

#3d Array

arr3 = np.array([
    [[1,2,3],
    [4,5,6]],
    [[1,2,3],
    [4,5,6]]
    
])

print(arr3.ndim)

```

Array Attributes

```python
#dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

#shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

#data types 
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#Size

print(arr1.size)
print(arr2.size)
print(arr3.size)   
```

```python
arr = np.array([1,2,3])

print(arr.dtype)

arr = np.array([1.5,2.4,3.2])
print(arr.dtype)

```

```python
arr = np.array([1,2,3.2])
#upcasted to float
print(arr.dtype)

#upcasted to string
arr = np.array([1,2,3.2,'hello'])
print(arr.dtype)
```

ndarray creation from existing data

```python
#List
lst = [10,20,30,40,40.5]

arr = np.array(lst, dtype=np.int32)

print(arr)
print(type(arr))
print(arr.dtype)

mixed_lst = [10, True, 'Hello', 10.5]

arr = np.array(mixed_lst)

print(arr)
print(type(arr))
print(arr.dtype)

matrix = [[1,2,3],
          [4,5,6]]

arr = np.array(matrix)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
tp1 = (10,20,30)

arr = np.array(tp1)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#set
st = {1,2,3}
arr = np.array(list(st))
print(arr.dtype)
arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#dictionary

dic = {'a' : 10, 'b' : 20, 'c' : 30}

keys = dic.keys()
print(keys)
values = dic.values()
print(values)

items = dic.items()
print(items)

#keys to ndarray
arr = np.array(list(keys))
print(arr.dtype)
# arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#items to ndarray
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

Arrray creation with range function

```python
# np.arange(first, end, step) ->similar to range function

# arr = np.arange(1, 10,1).reshape(3,3)
arr = np.arange(1, 10,1)
mat = arr.reshape(3,3)

print(arr)
print(mat)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.Linspace(start, end(include), koyta)

arr = np.linspace(1, 10,5)
# mat = arr.reshape(3,3)

print(arr)
# print(mat)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start,end(include),koyta)

arr = np.logspace(0,4,5,base=2)
print(arr)
```
