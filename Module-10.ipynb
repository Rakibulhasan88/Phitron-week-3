Using Numpy Array


```python
import numpy as np
arr1 = np.array([1,2,3])

arr1 = arr1 *5
print(arr1)
```

Ndarray

```python
import numpy as np

# 1d array

arr1 = np.array([1,2,3,4,5])

print(type(arr1))
print(arr1.ndim)

# 2d Array

arr2 = np.array([[1,2,3], #row1
                 [4,5,6] #row2
                ]) 

print(arr2.ndim)

#3d Array

arr3 = np.array([
    [[1,2,3],
    [4,5,6]],
    [[1,2,3],
    [4,5,6]]
    
])

print(arr3.ndim)

```

Array Attributes

```python
#dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

#shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

#data types 
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#Size

print(arr1.size)
print(arr2.size)
print(arr3.size)   
```
