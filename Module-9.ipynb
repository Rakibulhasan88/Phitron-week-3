```python
class Phone:
    category = "Electronics"

    # constructor
    def __init__(self, model, battery, camera, battery_percentage=100):
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage

    # methods
    def charge(self, hour):
        self.battery_percentage += hour

    def capture(self, photo):
        if self.battery_percentage <= 0:
            print("No charge")
        else:
            self.battery_percentage -= photo
            print("Photo Captured.")
            
class Cooling_Mechanism:
    def __init__(self,cooling_method):
        self.cooling_method = cooling_method
        
        
    #methods
    def cooling_on(self):
        print(f"the system is being cool by {self.cooling_method}")
```

```python
## Creating objects

apple = Phone("Iphone17",3000,40)

blueberry = Phone("B17",4000,30)

motorola = Phone("M-17",3000,35)

print(motorola.battery_percentage)

```

Single Inheritance

```python
# child / derived class
class SmartPhone(Phone):
    def __init__ (self, model, battery, camera, processor):
        super().__init__(model,battery,camera)
        self.peocessor = processor
        
    def charge(self, hour):
        print("fast charging in process")
        super().charge(hour)
```

```python
pro = SmartPhone("X",5000,100,"SnapDragon")
print(pro.model)

pro.charge(1)
```

Multiple Inheritance

```python
class SmartPhone_CoolingMode(SmartPhone,Cooling_Mechanism):
    def __init__(self,model, battery, camere,processor,cooling_method):
        SmartPhone.__init__(self,model,battery,camere,processor)
        Cooling_Mechanism.__init__(self,cooling_method)
```

```python
pro_cooling = SmartPhone_CoolingMode("Y",5000,100,"SD", "Nitrogen")

print(pro_cooling.peocessor) #SmartPhone class theke
print(pro_cooling.battery) #Phone class theke
print(pro_cooling.cooling_method) #Cooling_Machanism class er theke
print(pro_cooling.cooling_on()) #Cooling_Machanism class er theke
print(pro_cooling.charge(1)) #SmartPhone er modified charge jeta Phone class theke inherited
```

```python
class Camera:
    def __init__(self, name):
        self.name = name
    
    #method    
    def capture(self):
        print("A photo is captured")
```

```python
class Smart_Phone(Camera):
    def __init__(self,name,resulation):
        super().__init__(name)
        self.resulation = resulation
    
    #method overriding
    def capture(self):
        print("Photo is captured by a Phone")
        
class DSLR(Camera):
    def __init__(self, name,resulation):
        super().__init__(name)
        self.resulation = resulation
        
    def capture(self):
            print("Photo is captured DSLR")

class Drone(Camera):
    def __init__(self, name,resulation):
            super().__init__(name)
            self.resulation = resulation
            
    def capture(self):
            print("Photo is captured by Drone")
            
phone = Smart_Phone("Phone", 3000)
dslr = DSLR("DSLR", 200)
drone = Drone("Drone", 150)


phone.capture()
dslr.capture()
drone.capture()

```
