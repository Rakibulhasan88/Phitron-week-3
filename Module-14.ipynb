Filtering Data

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva","Rakib"],
    "City": ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix","H Los Angeles"],
    "Department": ["HR", "Finance", "IT", "Marketing", "Sales","HR"],
    "Salary": [70000, 80000, 90000, 75000, 85000,80000]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.loc[df['City'].str.contains("New")]
```

```python
df.loc[df['City'].str.contains("new",case = False)]
```

```python
# Start with loss
df.loc[df['City'].str.contains(r"^Los")]
```

```python
#end with rk
df.loc[df['Name'].str.contains(r"rk$")]

```

```python
# Name start with a vowel

df.loc[df['Name'].str.contains(r"^[AEIOUaeiou]")]
```

```python
#String contains New or Los
df.loc[df['City'].str.contains(r"New|Los")]
```

Handling duplicates Values

```python
data = {
    "Name" : ["Alice", "Bob", "Charlie", "David", "Eva","Rakib","Bob"],
    "City" : ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix","H Los Angeles","Los Angeles"],
    "Score" :[85, 90, 78, 92, 88, 95, 90]
}

df = pd.DataFrame(data)
df
```

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
#delete duplicate rows
df.drop_duplicates()
```

```python
#permanent delete
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
#column wise duplicate check
df.duplicated(subset=['Name'])
```

```python
#column wise duplicate delete
df.drop_duplicates(subset=['Name'], inplace=True)
```
